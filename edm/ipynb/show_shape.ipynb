```python
import os
import math
import numpy as np
import matplotlib.pyplot as plt

# ───────── 전역 스타일 ─────────
plt.rcParams.update({
    "font.size":        12,
    "axes.labelsize":   12,
    "xtick.labelsize":  10,
    "ytick.labelsize":  10,
    "legend.fontsize":  10,
    "axes.grid":        False,
    "axes.spines.top":  False,
    "axes.spines.right":False,
})

tab10   = plt.get_cmap('tab10').colors
c_shape = tab10[0:2]                     # pred / corr (2 색)

def plot_shape_grid(nfe_list, shape_dir="shape_dir", p=3,
                    n_rows=2, figsize=(18, 6)):
    """NFE 배열을 알아서 n_rows×n_cols 그리드로 시각화."""
    n_cols  = math.ceil(len(nfe_list) / n_rows)
    fig, axes = plt.subplots(n_rows, n_cols,
                             figsize=figsize,
                             sharey=True)
    axes = np.atleast_1d(axes).ravel()   # 1차원으로 평탄화

    for idx, (ax, nfe) in enumerate(zip(axes, nfe_list)):
        data = np.load(f"{shape_dir}/NFE={nfe},p={p}.npz")["optimal_log_shapes"]
        print('NFE: ', nfe)
        print('Predictor :', data[0, :])
        print('Corrector :', data[1, :])
        
        T    = data.shape[1]
        ax.plot(range(T), data[0], lw=1.6, marker="x",
                color=c_shape[0], label="predictor")
        ax.plot(range(T), data[1], lw=1.6, marker="x",
                color=c_shape[1], label="corrector")
        ax.set_title(f"NFE = {nfe}")
        ax.set_xticks([0, T-1])
        ax.set_xticklabels([0, T-1])
        ax.set_ylim(-2.2, 2.2)
        ax.grid(True, linestyle="--", alpha=0.6)
        if idx % n_cols == 0:
            ax.set_ylabel("shape")

    # 남은 빈 축 숨기기
    for ax in axes[len(nfe_list):]:
        ax.axis("off")

    handles, labels = axes[0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="lower center",
               bbox_to_anchor=(0.5, -0.05), ncol=2,
               frameon=True, edgecolor="black")
    fig.tight_layout(rect=[0, 0.05, 1, 1])
    plt.show()

# ───────── 사용 예시 ─────────
plot_shape_grid([5, 6, 8, 10, 12, 15, 20, 25, 30, 35, 40])
plot_shape_grid([5, 6, 8, 10, 12, 15, 20, 25, 30, 35, 40], p=4)
```
