```python
#!/usr/bin/env python3
import os
import numpy as np

for root, _, files in os.walk("."):
    for fname in files:
        if fname.endswith(".npz"):
            path = os.path.join(root, fname)
            with np.load(path, allow_pickle=True) as npz:
                data = {k: npz[k] for k in npz.files}
            
            if "optimal_log_scales" in data:
                data["optimal_log_shapes"] = data.pop("optimal_log_scales")
                np.savez(path, **data)
```
